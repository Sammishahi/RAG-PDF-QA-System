```python
#configuration
docs_dir_path = "C:\\Users\\HP\\Desktop\\rag-lang chain\\docs directory"
vector_store_path = "C:\\Users\\HP\\Desktop\\rag-lang chain\\vector_store"
collection_name = "doc_collection"
```

# load embedding model 
# it converts text into vector representation 
# using sentence transformers model

```python

embeddings = HuggingFaceEmbeddings()
```

```python
# directory loader to load all PDF files from the specified directory
from langchain_community.document_loaders import UnstructuredFileLoader
loader=DirectoryLoader(

    path=docs_dir_path,
    glob="./*.pdf",
    loader_cls=UnstructuredFileLoader,
    )
```

```python
#load the documents
documents=loader.load()
```

```python
print("Documents loaded successfully.")
print(type(documents))
```

```python
len(documents)
```

```python
#initializing the text splitter
text_splitter = CharacterTextSplitter(  
    chunk_size=1000,
    chunk_overlap=500,
)
```

```python
#splitting the documents into smaller chunks
text_chunks = text_splitter.split_documents(documents)
```

```python
len(text_chunks)
```

```python
text_chunks[0]
```

```python
# creating a vector store from the text chunks and embeddings
vector_store = Chroma.from_documents(
    documents=text_chunks,
    embedding=embeddings,
    persist_directory=vector_store_path,
    collection_name=collection_name
)
```
