```python
import os
import langchain
from langchain_huggingface import HuggingFaceEmbeddings     
from langchain_groq import ChatGroq
from langchain_classic.chains import RetrievalQA
```

```python
from dotenv import load_dotenv
```

#path of the data base

```python
#configuration
vector_store_path = "C:\\Users\\HP\\Desktop\\rag-lang chain\\vector_store"
collection_name = "doc_collection"
```

```python
#loading the embedding model - default model is "sentence-transformers/all-MiniLM-L6-v2"
embedding_model = HuggingFaceEmbeddings()
```

```python
# initialize groq llm 
llm = ChatGroq(
    model="llama-3.1-8b-instant",
    temperature=0.0,
)
```

```python
from langchain_chroma import Chroma
```

```python
vector_store = Chroma(
    collection_name=collection_name,
    embedding_function = embedding_model,
    persist_directory=vector_store_path
)
```

```python
retriever = vector_store.as_retriever()
```

```python
qa_chain = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff",
    retriever=retriever,
    return_source_documents=True
)
```

```python
query = "What is RCCGnet?"
response = qa_chain.invoke({"query": query})


#for source in response["source_documents"]:
    #print(source.metadata)
import textwrap

print(textwrap.fill(response["result"], width=80))
print("-"*50)
```
